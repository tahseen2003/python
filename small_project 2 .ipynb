first we will import pandas library.

```python
import pandas as pd
```

```python
# load dataset :
patients=pd.read_csv("hospital_patients.csv")
```

```python
# Show the datas
patients.head(10)
```

```python

patients.describe()
```

```python
# shows information about dataset
patients.info()
```

```python
# to shows missing null values
patients.isnull()
```

```python
# drop the duplicated values first.
patients.duplicated()
patients=patients.drop_duplicates(["patient_id"])
patients

```

```python
# write the names in sentence format
patients["name"]=patients["name"].str.strip().str.capitalize()
patients
```

```python
# distinct word used for male and female as there was multiplenames used for male and female.
patients["gender"].unique()
patients["gender"]=patients["gender"].replace({"M":"male","m":"male","F":"female","f":"female"})
patients["gender"]=patients["gender"].fillna("male")
patients
```

```python
# convert the age column to numeric format.
patients["age"]=pd.to_numeric(patients["age"],errors="coerce")
# remove the negative sign as the age must be positive.
patients["age"]=patients["age"].abs()
# fill the missing values using mean.
age_mean=patients["age"].mean()
patients["age"]=patients["age"].fillna(age_mean)
#round the values to int format from float 
patients["age"]=patients["age"].round().astype(int)
patients
```

```python
# the date have mixed and multiple format. so we will convert into one single format.
patients['admission_date']=pd.to_datetime(patients["admission_date"] ,format='mixed')
patients


```

```python
# convert Discharge_date into datetime format from string.
patients['discharge_date']=pd.to_datetime(patients['discharge_date'])
# we have 2 missing values and observing from other data there is 5 days difference between admission_date 
# discharge date so we fill the missing values by adding 5 days to discharge_date.
patients.loc[patients['discharge_date'].isna(),"discharge_date"] =(
    patients.loc[patients['discharge_date'].isna(),"admission_date"]+pd.Timedelta(days=5))
patients
```

```python
patients['department'].unique()
# set departments name into sentence format where first letter will be capital and other will be small.
patients['department']=patients['department'].str.strip().str.capitalize()
patients
```

```python
# we have don't idea about city and we cannot fill randomly or using mode so we set
# it to unknown.
patients['city'].fillna('unknown',inplace=True)
patients
```

```python
# there is dollar sign involoves so simply we remove the dollar sign.
patients['bill_amount']=patients['bill_amount'].str.replace("$",'',regex=False)
patients
```

```python
# the index was inconsistent 2 was missing so we reset the index.
patients.reset_index(drop=True,inplace=True)
patients
```

## finally we clean , manipulate the dataset for use.

```python
patients.to_csv("update_hospital.csv",index=False)
```
