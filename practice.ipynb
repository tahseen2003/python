## Series in pandas:
A Pandas Series is like a column in a table.

It is a one-dimensional array holding data of any type.
general syntax : Series() : first letter S must be capital.

```python
dic={
    "name":["tahseen","yasir","noman","haroon"],
    "roll-no":[18,21,17,19]
}
new=pd.Series(dic)
print(new)
```

### indexing in series:
indexing will be automitically start from 0 upto n: 
the value can be acceess through index number:
syntax : name [index number].
we can aslo give a random values to index:

```python
dic={
    "name":["tahseen","yasir","noman","haroon"],
    "roll-no":[18,21,17,19]
}
new=pd.Series(dic)
# accessing the insex 1 value:
print(new[1])
# taking a ist of three numbers:
list=[1,2,3]
# assiging index values x,y,z to values:
new=pd.Series(list,index=["x","y","z"])
print(new["x"])
```

### DataFrame in pandas:
Data sets in Pandas are usually multi-dimensional tables, called DataFrames.

Series is like a column, a DataFrame is the whole table.


```python
dic={
    "name":["tahseen","yasir","noman","haroon"],
    "roll-no":[18,21,17,19]
}
new=pd.DataFrame(dic)
print(new)
```

### accesing values in dataframes:
we will acces values through several ways but if we want to access 
one row in table then will use loc() function>
## syntax
loc[index number]

```python
dic={
    "name":["tahseen","yasir","noman","haroon"],
    "roll-no":[18,21,17,19],
    "subjects":["python","web dev","DBMS","digital marketting"]
}
new=pd.DataFrame(dic)
# we want to access the second row :
print(new.loc[2])
# but in this method the output will be in series form.

# similarly if we want t0 access multiple values at a same time.

print(new.loc[[0,2]])
#same as a series we can a give our own values as a index:
new=pd.DataFrame(dic,index=["a","b","c","d"])
print(new)
print (new.loc["c"])
```

## Empty Cells
Empty cells can potentially give you a wrong result when you analyze data.

Remove Rows
One way to deal with empty cells is to remove rows that contain empty cells.

This is usually OK, since data sets can be very big, and removing a few rows will not have a big impact on the result.

```python
# simple we output the data set:
print(file.to_string)
# we have a null cells in 18,23,28 position and we want to remove these rows:
# so we will use dropna() function.
# inplace = true : it means to returns the new update dataframe.
## file.dropna(inplace=True)
file
```

## replace Empty Values
Another way of dealing with empty cells is to insert a new value instead.

This way you do not have to delete entire rows just because of some empty cells.

The fillna() method allows us to replace empty cells with a value:

```python
# if we want to insert 300 in 18 row empty cell then will use:
file.fillna({"Duration":300})
file
```
