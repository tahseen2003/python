## Empty Cells
Empty cells can potentially give you a wrong result when you analyze data.

Remove Rows
One way to deal with empty cells is to remove rows that contain empty cells.

This is usually OK, since data sets can be very big, and removing a few rows will not have a big impact on the result.

```python
# simple we output the data set:
print(file.to_string)
# we have a null cells in 18,23,28 position and we want to remove these rows:
# so we will use dropna() function.
# inplace = true : it means to returns the new update dataframe.
## file.dropna(inplace=True)
file
```

## replace Empty Values
Another way of dealing with empty cells is to insert a new value instead.

This way you do not have to delete entire rows just because of some empty cells.

The fillna() method allows us to replace empty cells with a value:

```python
# if we want to insert 300 in 18 row empty cell then will use:
file.fillna({"Duration":300})
file
```

## Replace Using Mean, Median, or Mode
A common way to replace empty cells, is to calculate the mean, median or mode value of the column.

Pandas uses the mean() median() and mode() methods to calculate the respective values for a specified column:

```python
import pandas as pd
file2=pd.read_csv("data.csv")
# in this code the null values in calories column will be replace by its column mean.
x=file2["Calories"].mean()
x2=file2["Date"].mode()[0]
file2.fillna({"Calories":x, "Date":x2} ,inplace=True)
file2
```

## join function in pandas 
it is used to join more than one dataframe with each other.
there are 4 types of joins:
1.left join
2.right join
3.inner join
4.outer join
by default will be left join.

```python
dic1=pd.DataFrame({"A":[1,2,3,4,5],"B":[6,7,8,9,10]})
dic2=pd.DataFrame({"C":[11,12,13,14,15],"D":[16,17,18,19,20]})
# join both dataframe with each other
dic1.join(dic2)
# join specific column of one data frame to another
dic1.join(dic2["C"])

```

```python
# another example by giving indexes and having null values.
dic1=pd.DataFrame({"A":[10,20,30,40],"B":[50,60,70,80]},index=["a","b","c","d"])
dic2=pd.DataFrame({"C":[5,15,],"D":[25,35]},index=["a","b"])
dic1.join(dic2)
# this function will out all the left values as there are more or less values in right dataframe
dic2.join(dic1,how="left")
# this function will out all the right values as there are more or less values in left dataframe
dic2.join(dic1,how="right")
# this function will out the union between dataframe
dic2.join(dic1,how="outer")
# this function will out the intersection between dataframe
dic2.join(dic1,how="inner")
```

```python
# let suppose we have 2 data frame of students with different class. and the column name of roll no 
# and student are same and we want join then we will face an error.called 
# columns overlap but no suffix specified
# then we will suffix which is used to add suffix to column and then join as we see there as
# same columns name.
class5th=pd.DataFrame({"name":["yasir","kamran","ahmad","nawaz"],"roll no":[1,2,3,4]})
class6th=pd.DataFrame({"name":["mudasir","javed","luqman","zaryab"],"roll no":[10,11,12,13]})
class5th.join(class6th, how="left",rsuffix="_6th",lsuffix="_5th")
#lsuffix means left column suffix
#rsuffix means right column suffix
```

```python
# concat() : is used to join data frames but vertically not like join horizantially.
class5th=pd.DataFrame({"name":["yasir","kamran","ahmad","nawaz"],"roll no":[1,2,3,4]})
class6th=pd.DataFrame({"name":["mudasir","javed","luqman","zaryab"],"roll no":[10,11,12,13]})
result= pd.concat([class5th,class6th])
result
# if we want to ignore index which is repeating in this function. we will use ignore_index=true.
result= pd.concat([class5th,class6th],ignore_index=True)
result
```
