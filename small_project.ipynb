```python
# import pandas library.
import pandas as pd
```

we have a dataset which include some missing, wrong, wrong format values.

```python
# load a dataset 
df=pd.read_csv("dumy.csv")
df
```

```python
# we need information about data such as columns , rows, null values datatype etc.
df.info()
```

```python
# before applying we need to know as there as duplicated values or not.
df.duplicated()
```

```python
# to know which cell contain null values.
df.isnull()
```

```python
# there is no name in 8 location record so we will delete it.
df=df.drop(8)
df
# we need to remove duplicated values.
df=df.drop_duplicates()
df











```

```python
# in age column there is some values in string format so we convert it into numeric type.
df["age"]=df["age"].replace({"twenty":20,"twenty five":25})

# after that there is some values which do not make sense such as abc,?,missing etc.
# so we convert the column into numeric type in which the data which not make a sense become 
# null or missing. 
df["age"]=pd.to_numeric(df["age"],errors="coerce")
print(df["age"].dtype)
```

```python
# after that there is age with negative values which is wrong so we make it positive.
df["age"]=df["age"].abs()
# after that there is some people age is showing below 10 so we ind the mean and 
# replace the of which is less than 10 by its mean. 
x=df["age"].mean()
df.loc[df["age"]<10, "age"]=x
df
# the age which contain null values will be fill by its mean.
age=df["age"].mean()
df["age"]=df["age"].fillna(age).astype(int)
```

```python
# convert salary column into numeric format.
df["salary"]=pd.to_numeric(df["salary"],errors="coerce")
# filling null values in salary clumn by its mean.
salary=df["salary"].mean()
df["salary"]=df["salary"].fillna(salary).astype(int)
df
```

```python
# finding all unique values in gender column.
df["gender"].unique()
```

```python
# as there is only two values are wrong so replace by it correct name.
df["gender"]=df["gender"].replace({"M":"Male","F":"Female"})
```

```python
# as we was removed row 8 so the index was become inconsistent so we will make 
#it consistent again.
df.reset_index(drop=True,inplace=True)
df
```
